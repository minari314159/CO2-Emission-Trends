# Cleaning Pipelines 

### Load & Initial Inspection

```python
import pandas as pd
import numpy as np
```

```python
raw_df = pd.read_csv("../Data/ghg_raw.csv")
raw_df.head(10)
```

```python
print("\n───────────────────────────── INITIAL INSPECTION ─────────────────────────────")
print(f"\nShape:    {raw_df.shape[0]:,} rows × {raw_df.shape[1]} columns")
print(f"\nColumns:  {list(raw_df.columns)}")
print(f"\ndtypes:\n{raw_df.dtypes}")
print(f"\nBasic stats (numeric):\n{raw_df.describe()}")
```

### Missing Values & Duplicates

```python
null_counts = raw_df.isnull().sum()
null_pct = (raw_df.isnull().mean() * 100).round(2)
print(f"\nNull counts:\n{null_counts}")
print(f"\nNull %:\n{null_pct}")
```

```python
# Strategy: fill missing emissions with median of same country/industry/gas group
raw_df["emissions_mt"] = raw_df.groupby(["country", "industry", "gas_type"])["emissions_mt"].transform(lambda x: x.fillna(x.median()))
# Fallback: fill any remaining with overall median
raw_df["emissions_mt"] = raw_df["emissions_mt"].fillna(
    raw_df["emissions_mt"].median())

print(f"\nAfter imputation — remaining nulls: {raw_df.isnull().sum().sum()}")
```
